```python
#bai2
def bs(arr,x):
    l=0
    r=len(arr)-1
    while l<=r:
        m=(l+r)//2
        if arr[m]==x:
            return m+1
        elif arr[m]<x:
            l=m+1
        else:
            r=m-1
    return -1
a=[1,3,5,7,10,12,16,18]
f=int(input())
print(bs(a,f))
```

```python
#bai6
def mergesort(arr):
    if len(arr)==1:
        return arr
    mid=len(arr)//2
    l=mergesort(arr[:mid])
    r=mergesort(arr[mid:])
    return merge(l,r)
def merge(l,r):
    tmp=[]
    i=j=0
    while i<len(l) and j<len(r):
        if l[i]<r[j]:
            tmp.append(l[i])
            i+=1
        else:
            tmp.append(r[j])
            j+=1
    while i<len(l):
        tmp.append(l[i])
        i+=1
    while j<len(r):
        tmp.append(r[j])
        j+=1
    return tmp
lst=list(map(int,input().split()))
sorted_lst=mergesort(lst)
print(sorted_lst)
#-1 4 2 5 10 -8 12 3
```
