```python
#b7
class LightSwitch:
    def __init__(self):
        self.is_on = False
    def toggle(self):
        self.is_on = not self.is_on
    def status(self):
        return "ON" if self.is_on else "OFF"
n,l = int(input()),LightSwitch()
while n!=0:
    if input()=="toggle":
        l.toggle()
    n-=1
print(l.status())
```

```python
#b14 - interestaccount
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
            print(True)
        else:
            print(False)
    def process(self,arg):
        if arg[0] == 'deposit':
            self.deposit(int(arg[1]))
        else:
            self.withdraw(int(arg[1]))
    def __repr__(self):
        return f"Final balance: {self.balance:.2f}"
class InterestAccount(BankAccount):
    def __init__(self, owner, balance, rate):
        super().__init__(owner, balance)
        self.rate = rate
    def add_interest(self):
        self.deposit(self.balance*self.rate)

b = InterestAccount(input(),int(input()),float(input()))
b.add_interest()
b
#test: An 100 0.3
```

```python
#b15
class Account:
    def __init__(self, owner, number):
        self._owner = owner
        self._number = number
        self._balance = 0.0
    def deposit(self, amount):
        self._balance += amount
    def printStatement(self):
        print(f"Owner: {self._owner}\nAccount Number: {self._number}\nBalance: {self._balance}")
a = Account(input(), input())
n = int(input())
while n!=0:
    a.deposit(float(input()))
    n-=1
a.printStatement()
```

```python
#b16
class Animal:
    def __init__(self, name):
        self.name = name
    def __str__(self):
        return f"Animal: {self.name}"
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)
    def __str__(self):
        return f"Dog: {super().__str__()}"
typ,name = input(),input()
if typ.upper() == "A":
    print(Animal(name))
else:
    print(Dog(name))
```
